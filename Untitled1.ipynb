```python
import torch
print("CUDA available:", torch.cuda.is_available())
print("GPU name:", torch.cuda.get_device_name(0) if torch.cuda.is_available() else "No GPU")
import os
import numpy as np
from PIL import Image, ImageDraw, ImageFont
import xml.etree.ElementTree as ET
from torch.utils.data import random_split
from torch.utils.data import Dataset, DataLoader, random_split
import torchvision
import torchvision.transforms as T
from torchvision.models.detection.faster_rcnn import FastRCNNPredictor
from torchvision.transforms import functional as F
from torchvision.models.detection import fasterrcnn_resnet50_fpn, FasterRCNN_ResNet50_FPN_Weights
import glob
import json
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import pandas as pd
import time
import platform
import re
import random
from tqdm import tqdm
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, precision_score, recall_score, f1_score
import gc
from torchvision.utils import draw_bounding_boxes
from torchvision.transforms.functional import to_pil_image
```

```python
# Prepared dataset: Augmentation
class ParkingDataset(Dataset):
    def __init__(self, root_dir, transforms=None):
        self.root_dir = root_dir
        self.transforms = transforms
        self.image_paths = []
        self.annotation_paths = []
        for xml_file in glob.glob(os.path.join(root_dir, "**", "*.xml"), recursive=True):
            image_file = xml_file.replace(".xml", ".jpg")
            if os.path.exists(image_file):
                self.image_paths.append(image_file)
                self.annotation_paths.append(xml_file)

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        attempts = 0
        while attempts < 10:
            img_path = self.image_paths[idx]
            xml_path = self.annotation_paths[idx]
            #print(f"Próba {attempts + 1}: {xml_path}")
            img = Image.open(img_path).convert("RGB")
            tree = ET.parse(xml_path)
            root = tree.getroot()
            boxes = []
            labels = []
            for space in root.findall("space"):
                if "occupied" not in space.attrib:
                    continue
                occupied = int(space.attrib["occupied"])
                contour = space.find("contour")
                if contour is None:
                    continue
                points = contour.findall("point")
                if len(points) != 4:
                    continue
                x_coords = [int(pt.attrib["x"]) for pt in points]
                y_coords = [int(pt.attrib["y"]) for pt in points]
                xmin = min(x_coords)
                xmax = max(x_coords)
                ymin = min(y_coords)
                ymax = max(y_coords)
                boxes.append([xmin, ymin, xmax, ymax])
                labels.append(occupied)
            if len(boxes) > 0:
                break
            idx = (idx + 1) % len(self.image_paths)
            attempts += 1

        if len(boxes) == 0:
            raise RuntimeError(f"Brak poprawnych danych w 10 próbach. Ostatni: {xml_path}")

        labels = [label + 1 for label in labels]
        boxes = torch.as_tensor(boxes, dtype=torch.float32)
        labels = torch.as_tensor(labels, dtype=torch.int64)
        image_id = torch.tensor([idx])
        area = (boxes[:, 3] - boxes[:, 1]) * (boxes[:, 2] - boxes[:, 0])
        iscrowd = torch.zeros((len(labels),), dtype=torch.int64)
        target = {
            "boxes": boxes,
            "labels": labels,
            "image_id": image_id,
            "area": area,
            "iscrowd": iscrowd
        }
        if self.transforms:
            img = self.transforms(img)
        else:
            img = F.to_tensor(img)
        return img, target
```

```python
weights = FasterRCNN_ResNet50_FPN_Weights.DEFAULT
model = fasterrcnn_resnet50_fpn(weights=weights)
#model = fasterrcnn_resnet50_fpn(weights=None, weights_backbone=None)

# Change the number of classes (background + empty + occupied)
num_classes = 3
in_features = model.roi_heads.box_predictor.cls_score.in_features
model.roi_heads.box_predictor = FastRCNNPredictor(in_features, num_classes)
## !!! without weight=None
#transform = T.Compose([
#    T.ToTensor()
#])
## !!! with weight=weight
transform = weights.transforms()

data_path = '/content/drive/MyDrive/ParkFastRCNN//HasXML/PUCPR/'
save_path = "/content/drive/MyDrive/ParkFastRCNN/runs/trainPUCPR"
os.makedirs(save_path, exist_ok=True)

# this folder contains Cloudy, Rainy, Sunny folders, each with several subfolders and finally .txt, .jpg, .xml files
xml_files = glob.glob(os.path.join(data_path, "**", "*.xml"), recursive=True)
print(f"Number of XML files found in {data_path}: {len(xml_files)}")
print("Sample XML files:", xml_files[:5])
jpg_files = [f.replace(".xml", ".jpg") for f in xml_files]
jpg_exists = [os.path.exists(f) for f in jpg_files]
print(f"Number of JPG files corresponding to XML: {sum(jpg_exists)} out of {len(jpg_files)}")

dataset = ParkingDataset(root_dir=data_path, transforms=transform)
print(f"Number of samples in dataset: {len(dataset)}")
dataset_size = len(dataset)
train_size = int(0.8 * dataset_size)  # 80% train, 20% test
test_size = dataset_size - train_size

train_dataset, test_dataset = random_split(dataset, [train_size, test_size])
train_loader = DataLoader(train_dataset, batch_size=8, shuffle=True, collate_fn=lambda x: tuple(zip(*x)))
test_loader = DataLoader(test_dataset, batch_size=4, shuffle=False, collate_fn=lambda x: tuple(zip(*x)))

device = torch.device('cuda') if torch.cuda.is_available() else torch.device('cpu')
torch.cuda.empty_cache()
gc.collect()
model.to(device)

params = [p for p in model.parameters() if p.requires_grad]
optimizer = torch.optim.SGD(params, lr=0.005, momentum=0.9, weight_decay=0.0005)
lr_scheduler = torch.optim.lr_scheduler.StepLR(optimizer, step_size=3, gamma=0.1)

num_epochs = 3
best_loss = float('inf')

def evaluate_detection_model(model, data_loader, device, iou_threshold=0.5):
    model.eval()
    total_tp = 0
    total_fp = 0
    total_fn = 0
    with torch.no_grad():
        for images, targets in tqdm(data_loader, desc="Evaluation"):
            images = list(img.to(device) for img in images)
            outputs = model(images)
            for i in range(len(images)):
                pred_boxes = outputs[i]['boxes'].cpu().numpy()
                pred_labels = outputs[i]['labels'].cpu().numpy()
                gt_boxes = targets[i]['boxes'].cpu().numpy()
                gt_labels = targets[i]['labels'].cpu().numpy()
                tp, fp, fn = match_detections(pred_boxes, pred_labels, gt_boxes, gt_labels, iou_threshold)
                total_tp += tp
                total_fp += fp
                total_fn += fn
    precision = total_tp / (total_tp + total_fp) if (total_tp + total_fp) > 0 else 0
    recall = total_tp / (total_tp + total_fn) if (total_tp + total_fn) > 0 else 0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
    return precision, recall, f1


def compute_iou(boxA, boxB):
    # box format: [x1, y1, x2, y2]
    x_left = max(boxA[0], boxB[0])
    y_top = max(boxA[1], boxB[1])
    x_right = min(boxA[2], boxB[2])
    y_bottom = min(boxA[3], boxB[3])
    if x_right < x_left or y_bottom < y_top:
        return 0.0
    intersection_area = (x_right - x_left) * (y_bottom - y_top)
    area_A = (boxA[2] - boxA[0]) * (boxA[3] - boxA[1])
    area_B = (boxB[2] - boxB[0]) * (boxB[3] - boxB[1])
    iou = intersection_area / float(area_A + area_B - intersection_area)
    return iou

def match_detections(pred_boxes, pred_labels, gt_boxes, gt_labels, iou_threshold=0.5):
    matched_gt = set()
    true_positives = 0
    false_positives = 0

    for i, (pb, pl) in enumerate(zip(pred_boxes, pred_labels)):
        best_iou = 0
        best_gt_idx = -1
        for j, (gb, gl) in enumerate(zip(gt_boxes, gt_labels)):
            if j in matched_gt:
                continue
            if pl != gl:
                continue
            iou = compute_iou(pb, gb)
            if iou > best_iou:
                best_iou = iou
                best_gt_idx = j
        if best_iou >= iou_threshold:
            true_positives += 1
            matched_gt.add(best_gt_idx)
        else:
            false_positives += 1

    false_negatives = len(gt_boxes) - len(matched_gt)
    return true_positives, false_positives, false_negatives

precision_list = []
recall_list = []
f1_list = []

for epoch in range(num_epochs):
    print(f"Epoch {epoch+1}/{num_epochs}")
    model.train()
    epoch_loss = 0
    loop = tqdm(train_loader, desc="Training batches")
    for images, targets in loop:
        images = list(image.to(device) for image in images)
        targets = [{k: v.to(device) for k,v in t.items()} for t in targets]

        loss_dict = model(images, targets)
        losses = sum(loss for loss in loss_dict.values())

        optimizer.zero_grad()
        losses.backward()
        optimizer.step()

        epoch_loss += losses.item()
        loop.set_postfix(loss=losses.item())
    lr_scheduler.step()
    precision, recall, f1 = evaluate_detection_model(model, test_loader, device)
    precision_list.append(precision)
    recall_list.append(recall)
    f1_list.append(f1)
    print(f"Epoch {epoch+1} metrics — Precision: {precision:.4f}, Recall: {recall:.4f}, F1-score: {f1:.4f}")

    avg_loss = epoch_loss / len(train_loader)
    print(f"Średnia strata w epoce {epoch+1}: {avg_loss:.4f}")

    # Zapisz wagę ostatniej epoki
    torch.save(model.state_dict(), os.path.join(save_path, "model_last.pth"))

    # Zapisz wagę najlepszej epoki (jeśli strata się poprawiła)
    if avg_loss < best_loss:
        best_loss = avg_loss
        torch.save(model.state_dict(), os.path.join(save_path, "model_best.pth"))
        print(f"Zapisano najlepszy model po epoce {epoch+1} ze stratą {best_loss:.4f}")

plt.figure(figsize=(10, 5))
plt.plot(precision_list, label='Precision')
plt.plot(recall_list, label='Recall')
plt.plot(f1_list, label='F1-score')
plt.xlabel('Epoch')
plt.ylabel('Score')
plt.title('Precision, Recall, F1-score per Epoch')
plt.legend()
plt.grid(True)
plt.savefig(os.path.join(save_path, 'metrics_plot.png'))
plt.close()

# CSV
metrics_df = pd.DataFrame({
    'epoch': list(range(1, num_epochs + 1)),
    'precision': precision_list,
    'recall': recall_list,
    'f1_score': f1_list
})
metrics_df.to_csv(os.path.join(save_path, 'metrics.csv'), index=False)
```

```python
# FASTER R-CNN
detect_save_path = "/content/drive/MyDrive/ParkFastRCNN/runs/detectPUCPR"
os.makedirs(detect_save_path, exist_ok=True)

print(f"Save path created or already exists: {detect_save_path}")
test_set_size = len(test_loader.dataset)
print(f"Number of images in test set: {test_set_size}")

def save_predictions(images, outputs, save_dir, start_idx=0, score_threshold=0.5):
    model.eval()

    # Label definitions
    label_definitions = {
        0: ("0", "blue"),
        1: ("1", "red"),
        2: ("???", "gray"),
    }

    # Default font setup
    try:
        font = ImageFont.truetype("DejaVuSans-Bold.ttf", 14)
    except IOError:
        font = ImageFont.load_default()

    for idx, (img_tensor, output) in enumerate(zip(images, outputs)):
        scores = output['scores'].cpu()
        high_conf_idx = scores > score_threshold

        boxes = output['boxes'][high_conf_idx].cpu()
        labels = output['labels'][high_conf_idx].cpu()
        scores = scores[high_conf_idx]

        # Convert tensor to PIL image
        img_pil = to_pil_image(img_tensor.cpu()).convert("RGB")
        draw = ImageDraw.Draw(img_pil)

        for box, label, score in zip(boxes, labels, scores):
            x1, y1, x2, y2 = box.tolist()

            # Get label text and color
            label_id = label.item()
            label_text, color = label_definitions.get(label_id, (f"0 {label_id}", "blue"))

            # Add confidence with spacing
            label_text = f"{label_text}    {score:.2f}"

            # Draw rectangle
            draw.rectangle([x1, y1, x2, y2], outline=color, width=2)

            # Text position: above the box
            text_bbox = draw.textbbox((x1, y1), label_text, font=font)
            text_width = text_bbox[2] - text_bbox[0]
            text_height = text_bbox[3] - text_bbox[1]
            text_y = max(y1 - text_height - 2, 0)

            # Optional black background behind text
            draw.rectangle([x1, text_y, x1 + text_width, text_y + text_height], fill=(0, 0, 0))

            # Draw text
            draw.text((x1, text_y), label_text, fill=color, font=font)

        # Save image
        img_pil.save(os.path.join(save_dir, f"detected_{start_idx + idx}.png"))
```
